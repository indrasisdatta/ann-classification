```python
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np

corpus = [
    "PersonA likes icecream",
    "PersonB hates icecream",
    "PersonC likes icecream and coffee",
    "PersonD hates icecream and coffee"
]
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import nltk

stopwords_list = list(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
word_list = []
for sent in corpus:
    words = nltk.word_tokenize(sent)
    word_temp = []
    for word in words:
        if word not in stopwords_list:
            word_temp.append(lemmatizer.lemmatize(word))
    word_list.append((' ').join(word_temp))
    # print(lemmatizer.lemmatize(sent))
# print(word_list)
```

```python
cv = CountVectorizer(max_features=500, ngram_range=(1,2))
X = cv.fit_transform(word_list)
feature_names =  cv.get_feature_names_out()
print(X.toarray())
print("Feature names:", feature_names)
print("Vocabulary", cv.vocabulary_)
```

```python
word_counts = np.sum(X.toarray(), axis=0)
freq_words = dict(zip(feature_names, word_counts))
freq_words
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer 

tfidf = TfidfVectorizer(max_features=500, ngram_range=(1,2))
X = tfidf.fit_transform(word_list).toarray()
print(X)
print(tfidf.vocabulary_)
```
