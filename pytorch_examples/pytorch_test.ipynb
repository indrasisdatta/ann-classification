### AutoGrad

```python
# y = x**2, z = sin (y)     Find dz/dx

x = torch.tensor(3.0, requires_grad=True)
y = x**2
z = torch.sin(y)
print(x, y, z)
```

```python
z.backward()
x.grad
```

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
print(x,y,w,b)
```

```python
z = w * x + b 
z
```

```python
y_pred = torch.sigmoid(z)
y_pred
```

```python
from torch.nn import functional as F
loss = F.binary_cross_entropy(y_pred, y)
print(loss.item())
```

```python
loss.backward() 
print(w.grad, b.grad)
```
